# **MEMBUAT SINYAL DISKRIT**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n = np.arange(0, 11)

x1 = np.sin(0.5 * np.pi * n)

x2 = np.where(n < 5, 0, 1)

print("n =", n)
print("x1[n] =", x1)
print("x2[n] =", x2)

plt.stem(n, x1)
plt.title("x1[n] = sin(0.5πn)")
plt.xlabel('n')
plt.ylabel("x1[n]")
plt.grid(True)
plt.show()

plt.stem(n, x2)
plt.title("x2[n] (Unit Step)")
plt.xlabel("n")
plt.ylabel("x2[n]")
plt.grid(True)
plt.show()
```

Karakteristik dari x1[n]:


*   merupakan sinyal sinus diskrit
*   bersifat periodik / berulang


*   nilainya berosilasi / bolak-balik antara -1 dan 1

Karakteristik dari x2[n]:


*   merupakan sinyal unit step
*   nilai awalnya 0 kemudian berubah menjadi 1 pada *n* = 5


*   bersifat tidak periodik

# **UJI HOMOGENITAS**

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(0, 11)

x = np.sin(0.5 * np.pi * n)

plt.stem(n, x)

plt.title("Sinyal Input x[n]")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()

alpha_values = [0.5, 2, -1]

for alpha in alpha_values:
    T_alpha_x = 2 * (alpha * x)

    alpha_T_x = alpha * (2 * x)

    print(f"\n Alpha = {alpha}")
    print("Apakah sama ?",
          np.allclose(T_alpha_x, alpha_T_x))

    plt.plot(n, T_alpha_x, 'o-', label = 'T(αx)')
    plt.plot(n, alpha_T_x, 'x--', label = 'αT(x)')

    plt.title(f'Uji Homogenitas α = {alpha}')
    plt.legend()
    plt.grid(True)
    plt.show()

    print(f"T(αx) = {T_alpha_x}")
    print(f"αT(x) = {alpha_T_x}")
```

# **UJI ADDITIVITAS**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n = np.arange(0, 11)

x1 = np.sin(0.5 * np.pi * n)

x2 = np.where(n < 5, 0, 1)
```

```python
def T(x):
  return 2 * x

Tx1_plus_x2 = T(x1 + x2)

Tx1_plus_Tx2 = T(x1) + T(x2)

selisih = Tx1_plus_x2 - Tx1_plus_Tx2
```

```python
plt.stem(n, x1)
plt.title("x1[n]")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()

plt.stem(n, x2)
plt.title("x2[n]")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()

plt.stem(n, Tx1_plus_x2)
plt.title("T(x1 + x2)")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()

plt.stem(n, Tx1_plus_Tx2)
plt.title("T(x1) + T(x2)")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()

plt.stem(n, selisih)
plt.title("Selisih")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

```python
print("Apakah memenuhi additivitas ?")
print(np.allclose(Tx1_plus_x2, Tx1_plus_Tx2))

print("\n Selisih :")
print(selisih)
```

# **BANDINGKAN SISTEM LINIER DAN NON-LINIER**

```python
import numpy as np
import pandas as pd

n = np.arange(0, 11)
x = np.sin(0.5 * np.pi * n)

x1 = np.sin(0.5 * np.pi * n)
x2 = np.where(n < 5, 0, 1)

alpha = 2
```

```python
T1_left_h = 2 * (alpha * x)
T1_right_h = alpha * (2 * x)

print("========== T1(x) = 2x ==========")
print("\n UJI HOMOGENITAS")

print("T(αx) =")
print(T1_left_h)

print("\n αT(x) =")
print(T1_right_h)

hasil_h1 = np.allclose(T1_left_h, T1_right_h)

print("\n Memenuhi Homogenitas ?")
print(hasil_h1)

T1_left_a = 2 * (x1 + x2)

T1_right_a = 2 * x1 + 2 * x2

print("\n UJI ADDITIVITAS")

print("T(x1 + x2) =")
print(T1_left_a)

print("\n T(x1) + T(x2) =")
print(T1_right_a)

hasil_a1 = np.allclose(T1_left_a, T1_right_a)

print("\n Memenuhi Additivitas ?")
print(hasil_a1)

T2_left_h = (alpha * x) ** 2
T2_right_h = alpha * (x ** 2)

print("\n \n ========== T2(x) = x2 ==========")
print("\n UJI HOMOGENITAS")

print("T(αx) =")
print(T2_left_h)

print("\n αT(x) =")
print(T2_right_h)

hasil_h2 = np.allclose(T2_left_h, T2_right_h)

print("\n Memenuhi Homogenitas ?")
print(hasil_h2)

T2_left_a = (x1 + x2) ** 2
T2_right_a = (x1 ** 2) + (x2 ** 2)

print("\n UJI ADDITIVITAS")

print("T(x1 + x2) =")
print(T2_left_a)

print("\n T(x1) + T(x2) =")
print(T2_right_a)

hasil_a2 = np.allclose(T2_left_a, T2_right_a)

print("\n Memenuhi Additivitas ?")
print(hasil_a2)
```
